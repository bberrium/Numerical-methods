```python
import sympy as sp

def func(f, epsilon, a, b):
  n=0
  while (b-a) >= epsilon:
    n+=1
    x = sp.symbols('x')
    df = sp.diff(f(x), x)
    solution = sp.solve(df)

    print(float(solution[0]))

    x1 = a + (b-a)/2
    print("f(a) =", f(a))
    print("f(b) =", f(b))

    if f(x1) < 0:
      a = x1
    else:
      b = x1

    if f(a) * f(b) < 0:
        print(f"Opposite signs! Root is inside {[a, b]}")
    else:
        print("Same signs")


  print(f"х_приближение = {x1}, n={n}")

```

```python
def f(x):
    return x**3 - 3*x + 1
func(f, 0.1, 1, 2 )
```
