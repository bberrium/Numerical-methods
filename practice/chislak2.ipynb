```python
def func(g, x0, epsilon=1e-3, max_iter=100):
  x_prev = x0
  for i in range(1, max_iter+1):
    x_curr = g(x_prev)
    diff = abs(x_curr-x_prev)

    print(f"{i}: x = {x_curr}, |x_n-x_(n-1)| = {diff}")

    if diff<epsilon:
      return x_curr
    x_prev = x_curr

print("poisk kornya x1 na (0,1)")
g1 = lambda x: (x**3 + 1) / 3
x1 = func(g1, x0=0, epsilon=1e-3)

print("poisk kornya x2 na (1, 2) cherez stepennuyu formu")
g2 = lambda x: (3*x - 1) ** (1/3)
x2 = func(g2, x0=1.5, epsilon=1e-3)

print("poisk kornya x2 na (1, 2) cherez parametr alpha=5")
f=lambda x: x**3 - 3*x + 1
g_alpha = lambda x: x - f(x)/5
x2_alpha = func(g_alpha, x0=1.0, epsilon=1e-3)
```
